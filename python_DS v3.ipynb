```python
text = "Python"

# Forward indexing
print(text[0])  # Output: P (First character)
print(text[2])  # Output: t

# Negative indexing
print(text[-1]) # Output: n (Last character)
print(text[-2]) # Output: o
```

```python
text = "Python Programming"

# Extract "Python" (Indices 0 to 5)
print(text[0:6])   # Output: Python

# Shortcut: If you leave the start blank, it starts at 0
print(text[:6])    # Output: Python

# Shortcut: If you leave the end blank, it goes to the very end
print(text[7:])    # Output: Programming

# Reverse a string using a negative step
print(text[::-1])  # Output: gnimmargorP nohtyP
```

```python
text = "    python code  "

print(text.upper())      # Output: "  PYTHON CODE  " (All caps)
print(text.strip())      # Output: "python code" (Removes empty spaces at start/end)
print(text.replace("code", "programming")) # Output: "  python programming  "

# Checking contents
print("programming" in text)  # Output: True (Checks if a word exists inside)
```

**7. String Formatting (F-Strings)The easiest way to inject variables inside a piece of text is by using f-strings. Just put an f before the opening quote and wrap your variables in {} brackets.**

```python
name = "Amna"
age = 26

# The modern, clean way to format text
message = f"Hello, my name is {name} and I am {age} years old."
print(message)
# Output: Hello, my name is Alex and I am 25 years old.
```

```python
team = [('shahid', 20, 'center'),
        ('Ana', 22, 'point guard'),
        ('Nabi', 27, 'power forward')]
```

```python
def player_position(players):
  result=[]
  for name, age, position in players:
    result.append('Name: {:>19} \nAge: {:>17} \nPosition: {:>15}\n'.format(name,age, position))
    return result
```

```python
for player in player_position(team):
  print(player)
```

Enumerate Function

```python
marks = [12, 56, 32, 98, 12, 45, 1, 4]


for index, mark in enumerate(marks):
  print(index, mark)
  if(index == 3):
    print("helo", "do awesome!")
```
